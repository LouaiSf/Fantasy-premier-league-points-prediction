```python
import pandas as pd

# Load the dataset
df = pd.read_csv('all_seasons_data.csv')

# find all the values of gameweek column that are greater than 38
wrong_gw_values = df[df['GW'] > 38]['GW'].unique()
print(f"Gameweek values greater than 38: {wrong_gw_values}")
```

```python
# for gameweek values greater than 38, reduce them by 9
for gw in wrong_gw_values:
    df.loc[df['GW'] == gw, 'GW'] = gw - 9
    print(f"Corrected gameweek {gw} to {gw - 9}")
# Save the corrected dataset
df.to_csv('all_seasons_data.csv', index=False)
```
